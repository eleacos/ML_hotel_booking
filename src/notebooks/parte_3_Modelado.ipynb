# Hotel Booking: cómo afectan las cancelaciones a los hoteles

Las cancelaciones de reservas representan un desafío para la industria hotelera, ya que generan pérdidas en los ingresos, provocan una mala planificación en la ocupación y derivan en sobrecostes operativos. Estas cancelaciones pueden estar relacionadas con factores como políticas de cancelación flexibles, el uso de múltiples plataformas de reserva o la ausencia de penalizaciones por cancelar en el último minuto, entre otras.

El dataset analizado recoge reservas realizadas entre el año 2015 y 2017, de las que aproximadamente un 37% son reservas canceladas, lo que indica la importancia de este fenómeno durante este periodo. El objetivo de este estudio es desarrollar un modelo de clasificación binaria capaz de predecir con antelación la probabildiad de cancelación de una reserva, permitiendo a los hoteles elaborar estrategias de actuación que reduzcan el impacto económico y operativo de las cancelaciones, apoyándose en datos históricos.

Este problema se puede abordar con Machine Learning porque disponemos de un historial de reservas con múltiples variables predictoras (tipo de hotel, tiempo de anticipación, número de adultos, plataforma de reserva, etc.) y un target binario (cancelada/no cancelada). Existen patrones en los datos que los algoritmos de clasificación supervisada pueden aprender para predecir futuras cancelaciones.

Se propone un enfoque de clasificación supervisada, porque el objetivo es predecir una variable binaria conocida a partir de características históricas. Este enfoque permite estimar probabilidades de cancelación para nuevas reservas y, por tanto, guiar decisiones operativas. Algoritmos como Random Forest, XGBoost o regresión logística son adecuados dado que manejan variables categóricas y numéricas, toleran relaciones no lineales y permiten interpretar la importancia de las variables.

## Análisis temporal para detectar estacionalidad:
Nuestro target puede verse afectado por las variables de tiempo que tenemos. Veamos la relación que tienen: 

```python
# distribución de cancelaciones por año:
X_train.arrival_date_year.value_counts()
```

```python
# Agrupamos por años y meses y calculamos la media de las cancelaciones, para establecer temporalidad:
df_train = X_train.copy()
df_train["is_canceled"] = y_train

cancel_by_month = (df_train.groupby(["arrival_date_year", "arrival_date_month"])["is_canceled"].mean() * 100).sort_index()
print(cancel_by_month)
```

- Inicialmente, hay aproximadamente 3 veces más cancelaciones en el 2016 que en el 2015.
- Sin embargo, al revisar la media de las cancelaciones por meses y años, vemos que no tenemos información de todos los meses de los años:
    - del 2015, tenemos información de 6 meses, de julio a diciembre.
    - del 2016, tenemos información de todos los meses del año.
    - del 2017 tenemos información de 8 meses, de enero a agosto solamente. 
- A partir de esto, vemos que en el 2015 el mayor mes de cancelación fue en julio; en el 2016, fue en octubre; en 2017 fue mayor y similar en abril, mayo y junio.

#### Veámoslo gráficamente con un heatmap:

```python
# ordenamos los meses para que se visualicen en orden temporal y no alfabético:
month_order = [
    "January", "February", "March", "April", "May", "June",
    "July", "August", "September", "October", "November", "December"
]
# pd.Categorical lo usamos para convertir la col arrival_date_month en un tipo categórico ordenado, en vez de str ya que los ordena alfabéticamente.
df_hotel["arrival_date_month"] = pd.Categorical(
    df_hotel["arrival_date_month"],
    categories=month_order,
    ordered=True
)
```

```python
# Creamos la tabla pivote para visualizar la media de cancelaciones por año y mes, con los meses ordenados:
pivot = (df_hotel.groupby(["arrival_date_year", "arrival_date_month"])["is_canceled"].mean().unstack() * 100)
# unstack convierte la segunda col arrival_date_month en columnas y deja los años como filas, y 
# resulta una matriz tipo tabla con años vs meses
```

```python
# Dibujamos el heatmap:
plt.figure(figsize=(12,6))

sns.heatmap(
    pivot, # usa la tabla pivote
    annot=True, # muestra los valores numéricos dentro de cada celda
    fmt=".1f", # con un decimal
    cmap="YlOrRd", # escala de colores amarillo a rojo
    linewidths=0.5 # separación entre celdas
)

plt.title("Porcentaje de cancelación por año y mes")
plt.ylabel("Año")
plt.xlabel("Mes")
plt.show();
```

### Resultados visuales:
- En el 2015, hubo más cancelaciones en julio, agosto y septiembre, temporada de verano.
- En el 2016, la tasa de cancelaciones se concentró en abril, junio y octubre, siendo octubre el mes con más cancelaciones, no hay una temporada clara.
- En el 2017, se concentraron las cancelaciones entre abril, mayo y junio, estando bastante igualadas en porcentaje, en la temporada de primavera

Selección de variables para modelado  
Para el modelado se creó un dataset reducido (df_model) con un conjunto de variables seleccionadas (cols_to_keep). Esta selección se realizó para:

- Evitar fuga de información, excluyendo variables que podrían depender de eventos posteriores a la reserva.  
- Mantener variables con sentido predictivo y coherencia de negocio, relacionadas con condiciones de la reserva (precio, antelación, tipo de depósito, canal/segmento), historial del      cliente y características operativas;  
- Simplificar el modelo y reducir ruido, favoreciendo una mejor generalización.

```python
# Nos quedamos únicamente con las columnas definidas
df_model = df[cols_to_keep].copy()  

df_model.head()
```

Tratamiento de Missing Values

```python
# Agrupación por país
def group_country(country):
    europe = ["PRT","GBR","ESP","FRA","DEU","ITA"]
    america = ["USA","BRA","CAN"]
    asia = ["CHN","IND","JPN"]

    if country in europe:
        return "Europe"
    elif country in america:
        return "America"
    elif country in asia:
        return "Asia"
    else:
        return "Other"

df_model["country_grouped"] = df_model["country"].apply(group_country)
df_model = df_model.drop(columns=["country"])
```

```python
df_model.isna().sum()
```

### **FEATURES ENGINEERING**  
TRANSFORMACIONES


```python
# Transformaciones logarítmicas

import numpy as np  # Librería para operaciones matemáticas

# --- ADR ---
df_model["adr"] = df_model["adr"].clip(lower=0)  # Evitamos valores negativos (si los hubiera)
df_model["adr_log"] = np.log1p(df_model["adr"])  # Aplicamos log(1+x) para reducir asimetría
df_model = df_model.drop(columns=["adr"])        # Eliminamos la variable original para evitar duplicidad

# --- Lead Time ---
df_model["lead_time_log"] = np.log1p(df_model["lead_time"])  # Transformación log para reducir cola larga

# --- Previous Cancellations ---
df_model["previous_cancellations"] = np.log1p(
    df_model["previous_cancellations"]
)  # Transformación log para suavizar valores extremos
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(15,4))

# ADR
plt.subplot(1,3,1)
sns.histplot(df_model["adr_log"], bins=50)
plt.title("ADR (log transformado)")

# Lead Time
plt.subplot(1,3,2)
sns.histplot(df_model["lead_time_log"], bins=50)
plt.title("Lead Time (log transformado)")

# Previous Cancellations
plt.subplot(1,3,3)
sns.histplot(df_model["previous_cancellations"], bins=30)
plt.title("Previous Cancellations (log transformado)")

plt.tight_layout()
plt.show()
```

Se aplicaron transformaciones logarítmicas únicamente a aquellas variables numéricas que presentan asimetría positiva significativa y rango amplio de valores (adr, lead_time, previous_cancellations).
No se transformaron variables discretas de bajo rango (adults, booking_changes, required_car_parking_spaces, total_of_special_requests) ya que el logaritmo no aporta mejora en términos de linealidad ni reducción de outliers.

**Nuevas Features**

```python
# Cliente con historial de cancelación
# Captura el patrón: “cliente reincidente”
df_model["had_previous_cancel"] = (df_model["previous_cancellations"] > 0).astype(int) 

# Ratio de cancelaciones históricas 
# Representa comportamiento real del cliente 
df_model["cancel_ratio"] = (
    df_model["previous_cancellations"] /
    (df_model["previous_cancellations"] + df_model["previous_bookings_not_canceled"] + 1)     
)

# Cliente recurrente coherente 
# Cliente repetidor sin historial de cancelación → baja probabilidad de cancelar
df_model["loyal_customer"] = (
    (df_model["is_repeated_guest"] == 1) &
    (df_model["previous_cancellations"] == 0)
).astype(int)

# Interacción Lead Time × ADR
# Clientes que reservan con mucha antelación y ADR alto pueden comportarse distinto.
df_model["lead_time_x_adr"] = df_model["lead_time_log"] * df_model["adr_log"] 

# Nivel de solicitudes especiales
# En vez de variable 0–5, crear categoría alta demanda: Porque 2 o más suele cambiar comportamiento.
df_model["many_special_requests"] = (
    df_model["total_of_special_requests"] >= 2
).astype(int)

# Parking como indicador premium
# Variable binaria más estable para nuestra baseline
df_model["needs_parking"] = (
    df_model["required_car_parking_spaces"] > 0
).astype(int)

# lead time alto
# Clientes que reservan con mucha antelación
df_model["long_lead"] = (
    df_model["lead_time"] > df_model["lead_time"].median()
).astype(int)

# Reservas no definidas 
df_model["market_segment"] = df_model["market_segment"].replace("Undefined", "Other")
```

Con el objetivo de mejorar la capacidad predictiva del modelo, se generaron nuevas variables derivadas a partir de las características originales del dataset. Estas transformaciones permiten capturar patrones de comportamiento del cliente y relaciones entre variables que pueden influir en la probabilidad de cancelación de una reserva.

En particular, se crearon variables relacionadas con el historial de cancelaciones del cliente (had_previous_cancel, cancel_ratio), la fidelidad del cliente (loyal_customer), interacciones entre variables relevantes (lead_time_x_adr), así como variables binarias que representan ciertas características de la reserva (many_special_requests, needs_parking, long_lead).

Estas nuevas variables permiten enriquecer la información disponible para el modelo y mejorar su capacidad para identificar patrones asociados a cancelaciones.

Se separa la variable objetivo (is_canceled) del conjunto de variables predictoras para iniciar el proceso de modelado.

**TRAIN/TEST SPLIT**

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

print("Train size:", X_train.shape)
print("Test size:", X_test.shape)
```

Se realiza una división estratificada para conservar la proporción de cancelaciones en ambos conjuntos.

**SEPARAR VARIABLES NUMÉRICAS Y CATEGÓRICAS**

```python
num_cols = X_train.select_dtypes(include=["int64", "float64"]).columns
cat_cols = X_train.select_dtypes(include=["object"]).columns

print("Numéricas:", len(num_cols))
print("Categóricas:", len(cat_cols))
```

**CONSTRUIMOS PIPELINES INTERNOS**

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("num", numeric_pipeline, num_cols),
    ("cat", categorical_pipeline, cat_cols)
])
```

📘 Tratamiento de variables mediante Pipeline  

Con el objetivo de garantizar un flujo de modelado robusto y evitar fuga de información, se construyen pipelines diferenciados para variables numéricas y categóricas.

Para variables numéricas se aplica:  
- Imputación mediante mediana, por su robustez frente a valores extremos.  
- Escalado estándar (StandardScaler) para normalizar la distribución y facilitar la convergencia del modelo.

Para variables categóricas se aplica:  
- Imputación mediante la categoría más frecuente.  
- Codificación One-Hot para transformar variables textuales en formato numérico.  

Estas transformaciones se integran mediante un ColumnTransformer, permitiendo que cada tipo de variable reciba el tratamiento adecuado dentro de un único flujo reproducible.

Este enfoque asegura que todas las transformaciones se ajusten únicamente con los datos de entrenamiento, evitando contaminación del conjunto de test.

**MODELADO BASELINE** - Logistic Regression
- Preprocesado  
- Modelado  

```python
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

log_model = Pipeline([
    ("preprocessing", preprocessor),
    ("model", LogisticRegression(
        max_iter=1000,
        class_weight="balanced",
        random_state=42
    ))
])
```

Se selecciona la Regresión Logística como modelo baseline por tratarse de un problema de clasificación binaria y por su equilibrio entre simplicidad, estabilidad e interpretabilidad.  
Este modelo permite establecer una referencia clara de rendimiento antes de introducir modelos más complejos como Random Forest.  
Además, su naturaleza lineal facilita la interpretación del impacto de las variables en la probabilidad de cancelación.

- Entrenamos

```python
log_model.fit(X_train, y_train)
```

- Evaluamos

```python
from sklearn.metrics import balanced_accuracy_score, classification_report, roc_auc_score

y_pred = log_model.predict(X_test)
y_proba = log_model.predict_proba(X_test)[:, 1]

print("Balanced Accuracy:", balanced_accuracy_score(y_test, y_pred))
print("ROC-AUC:", roc_auc_score(y_test, y_proba))
print("\nClassification Report:\n")
print(classification_report(y_test, y_pred))
```

**DEFINIR VARIOS MODELOS**  

Vamos a comparar 4:  
Logistic Regression: se utiliza como modelo baseline por su simplicidad, interpretabilidad y buen rendimiento en problemas de clasificación binaria.  
Random Forest: se selecciona por su capacidad para capturar relaciones no lineales y reducir el overfitting mediante el uso de múltiples árboles de decisión.  
Gradient Boosting: se incluye porque es un modelo de ensemble secuencial que mejora iterativamente los errores, lo que suele ofrecer alto rendimiento en problemas complejos.  
KNN (K-Nearest Neighbors): se utiliza para comparar con un enfoque basado en similitud entre observaciones, donde la clasificación depende de los vecinos más cercanos en el espacio de características.  


```python
# RANDOM FOREST
from sklearn.pipeline import Pipeline
from sklearn.metrics import balanced_accuracy_score, roc_auc_score 

from sklearn.ensemble import RandomForestClassifier

rf = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("model", RandomForestClassifier(
        n_estimators=300,
        random_state=42,
        class_weight="balanced"
    ))
])

rf.fit(X_train, y_train)

y_pred_rf = rf.predict(X_test)
y_proba_rf = rf.predict_proba(X_test)[:, 1]

print("Random Forest")
print("Balanced Accuracy:", balanced_accuracy_score(y_test, y_pred_rf))
print("ROC-AUC:", roc_auc_score(y_test, y_proba_rf))
```

```python
#  GRADIENT BOOSTING
from sklearn.ensemble import GradientBoostingClassifier

gb = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("model", GradientBoostingClassifier(random_state=42))
])

gb.fit(X_train, y_train)

y_pred_gb = gb.predict(X_test)
y_proba_gb = gb.predict_proba(X_test)[:, 1]

print("Gradient Boosting")
print("Balanced Accuracy:", balanced_accuracy_score(y_test, y_pred_gb))
print("ROC-AUC:", roc_auc_score(y_test, y_proba_gb))
```

```python
# KNN
from sklearn.neighbors import KNeighborsClassifier

knn = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("model", KNeighborsClassifier(n_neighbors=15))
])

knn.fit(X_train, y_train)

y_pred_knn = knn.predict(X_test)
y_proba_knn = knn.predict_proba(X_test)[:, 1]

print("KNN")
print("Balanced Accuracy:", balanced_accuracy_score(y_test, y_pred_knn))
print("ROC-AUC:", roc_auc_score(y_test, y_proba_knn))
```

**Validación Cruzada**  
Ahora queremos estimar el rendimiento real del modelo, probando distintas versiones del mismo modelo para encontrar la mas equilibrada

```python

# PASO 1: Setup GridSearch


from sklearn.model_selection import StratifiedKFold, GridSearchCV   # CV estratificada + búsqueda grid
from sklearn.ensemble import RandomForestClassifier                 # Modelo RF
from sklearn.pipeline import Pipeline                               # Encadenar preprocesado + modelo

# 1) Validación cruzada estratificada (mantiene proporción de clases en cada fold)
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)     # 5 folds estratificados

# 2) Pipeline: mismo preprocesado para todos los modelos
rf_pipe = Pipeline(steps=[
    ("preprocessor", preprocessor),                                  # OneHot + Scaler (lo que ya definiste)
    ("model", RandomForestClassifier(
        random_state=42,                                            # Reproducible
        class_weight="balanced",                                    # Compensa desbalance
        n_jobs=-1                                                   # Usa todos los cores
    ))
])

# 3) Grid de hiperparámetros (pequeño para empezar y no tardar demasiado)
param_grid = {
    "model__n_estimators": [200, 400],                               # Nº de árboles
    "model__max_depth": [8, 12, 16],                              # Profundidad máxima
    "model__min_samples_split": [5, 10],                             # Mín. muestras para dividir
    "model__min_samples_leaf": [2, 5],                               # Mín. muestras en hoja
    "model__max_features": ["sqrt", "log2"]                          # Nº de features por split
}
```

**Preparación para la optimización de hiperparámetros** 

Antes de ejecutar la búsqueda de hiperparámetros, se definió:  
 - Una validación cruzada estratificada (StratifiedKFold) para obtener estimaciones robustas del rendimiento.  
 - Un Pipeline que integra el preprocesado y el modelo, evitando fuga de información durante la validación cruzada.  
 - Una rejilla de hiperparámetros que controla la complejidad y capacidad del modelo Random Forest.  
Este procedimiento garantiza que la selección del modelo final esté basada en evaluación rigurosa y reproducible.  

**OPTIMIZACIÓN HIPERPARÁMETROS**

```python
#  Ejecutar GridSearch

grid = GridSearchCV(
    estimator=rf_pipe,                     # pipeline con preprocesado + RandomForest
    param_grid=param_grid,                # grid de hiperparámetros que definimos
    scoring={
        "bal_acc": "balanced_accuracy",    # métrica principal
        "roc_auc": "roc_auc"               # métrica secundaria
    },
    refit="bal_acc",                       # el mejor modelo se elige por balanced accuracy
    cv=cv,                                 # validación cruzada estratificada
    n_jobs=-1,                             # usa todos los cores
    verbose=1                              # muestra progreso
)

# Ejecutar la optimización
grid.fit(X_train, y_train)
```

Que queremos conseguir con la optimización de hiperparámetros:  
1️⃣ Probar todas las combinaciones de param_grid  
2️⃣ Para cada combinación hacer 5-fold cross validation  
3️⃣ Calcular balanced accuracy y roc-auc  
4️⃣ Guardar el mejor modelo  

**Mejor modelo encontrado**

```python
print("Mejores hiperparámetros:")
print(grid.best_params_)

print("\nMejor balanced accuracy en CV:")
print(grid.best_score_)
```
